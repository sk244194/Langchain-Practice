Using Groq

```python
from langchain.chat_models import init_chat_model

# Search on google the name of model and model provider

llm = init_chat_model(
    model="llama-3.3-70b-versatile",
    model_provider="groq", 
    temperature=0
    # temperature tells how creative ai should be if closer to 1 then more creative but can create wrong answers
)

response= llm.invoke("Who created you")
# response
response.content
```

Another Way of doing it

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

res = llm.invoke("How are you")
res.content
```

Streams

This is used to print data while it is generating

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

# res = llm.stream("How are you")
# res
# This will return a generator object 

# To print data
for chunk in llm.stream("How are you"):
    print(chunk.text,end="")
```

Batch

If i wanted to pass multiple queries to llm and want it to run it parallely then batch is used

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

res = llm.batch(["How are you",
                 "What is Claude's Next Project"

]
# ,config =  {'max_concurrency':5 -> This tells how much queries we want it to run parallely}
)

for response in res:
    print(response)
```
