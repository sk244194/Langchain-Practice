Tools -> Models can request to call tools that perform tasks such as fetching data from database, searching the web
or running code. Tools are pairing of a schema and a function ro execute


Every LLM has a cutoff training date which means if i ask it something after that then it will give wrong result

```python
import langchain

import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]= os.getenv("groq_api_key")
```

Agent 

```python
from langchain.agents import create_agent
from langchain.tools import tool

@tool
def get_job(job):
    """ Finding jobs """
    return f'how {job} is going to be affected by AI'

agent = create_agent(
    model="groq:qwen/qwen3-32b",
    tools=[get_job],
    system_prompt="You are a coder looking for a job"
)

response = agent.invoke({"messages":[{"role":"user","content":"how cybersecurity is going to be affected"}]})

response["messages"][-1].content
#There is a difference in agent to print result compared to how it was being printed using response.content previously
```

Another Way of using Tools but agent one is more preferred

```python
from langchain.tools import tool
from langchain.chat_models import init_chat_model

@tool
def get_job(job):
    """ Finding jobs """
    return f'how {job} is going to be affected by AI'

llm = init_chat_model("groq:qwen/qwen3-32b")

llm_with_tool = llm.bind_tools([get_job])

response = llm_with_tool.invoke("Is Data Analysts going to be affected by AI")

# Sometimes AI would prefer not to use the tool then this will return an empty array
response.tool_calls
```

```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_tavily import TavilySearch

load_dotenv()
 
os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")


search = TavilySearch(max_results=3)

result = search.invoke({
    "query": "Latest AI news"
})

result
```
